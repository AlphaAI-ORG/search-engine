```python
# RecursiveCharacterTextSplitter 청킹

from langchain_text_splitters import RecursiveCharacterTextSplitter

# text_splitter = RecursiveCharacterTextSplitter(
#     chunk_size=500,       
#     chunk_overlap=200,    
#     length_function=len,  
#     is_separator_regex=False, 
# )

text_splitter = RecursiveCharacterTextSplitter(
    separators=[
        "\n\n",   # 두 개의 개행 문자 (문단 분리)
        "\n",     # 단일 개행 문자 (줄바꿈)
        " ",      # 공백 (단어 단위 분리)
        ".",      # 마침표
        ",",      # 쉼표
        "\u200b",  # Zero-width space (보이지 않는 공백 문자)
        "\uff0c",  # Fullwidth comma (중국어, 일본어에서 사용)
        "\u3001",  # Ideographic comma (중국어, 일본어에서 사용)
        "\uff0e",  # Fullwidth full stop (중국어, 일본어에서 사용)
        "\u3002",  # Ideographic full stop (중국어, 일본어에서 사용)
        "",       # 마지막 분할 기준 (기본적으로 아무 구분자가 없을 경우)
    ],
    chunk_size=500,       # 기존 설정 유지
    chunk_overlap=200,    # 기존 설정 유지
    length_function=len,  # 기존 설정 유지
)
# PDF에서 로드한 데이터를 텍스트 청크로 분할
texts = text_splitter.split_documents(pages)

print(f"{texts[0].metadata}") # 첫 번째 청크의 메타데이터 
print(texts[0].page_content) # 첫 번째 청크의 내용
print("-"*100)
print(f"{texts[1].metadata}") # 두 번째 청크의 메타데이터
print(texts[1].page_content) # 두 번째 청크의 내용

# TODO : semantic chunking 시도
```

```python
# %pip install langchain_openai
from langchain_openai import OpenAIEmbeddings
embeddings_model = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
author_info = {"아빠와 딸의 주식 투자 레슨": "대니얼 타운 외 1명"}
print(author_info["아빠와 딸의 주식 투자 레슨"])
```

```python
import os
import glob
from tqdm import tqdm
from langchain_community.document_loaders import PyPDFLoader

def process_pdfs():
   # PDF 파일 목록 가져오기
   pdf_files = glob.glob("./data/*.pdf")
   print(f"처리할 PDF 파일 수: {len(pdf_files)}")

   # 벌크 인덱싱을 위한 데이터 준비
   bulk_data = []
   
   for book_id, pdf_file in enumerate(pdf_files): #tqdm(pdf_files, desc="모든 PDF 파일"):
       # PDF 로드
       loader = PyPDFLoader(pdf_file)
       pages = loader.load()  # 비동기 대신 동기 방식으로 변경
       
       #import ipdb; ipdb.set_trace()
       # 텍스트 청킹
       
       texts = text_splitter.split_documents(pages)
       
       # 임베딩 생성 및 벌크 데이터 준비
       for i, text in tqdm(enumerate(texts), total=len(texts), desc="파일 처리 중"):
           # 임베딩 생성
           embedding = embeddings_model.embed_query(text.page_content) # small-1536
           embedding_1024 = embedding[:1024]  # 1024 차원으로 축소
           
           # 메타데이터 추출
           metadata = text.metadata
           book_name = os.path.basename(pdf_file).replace('.pdf', '').strip()  # 공백 제거
           
           author = author_info.get(book_name, "")
           
           # 다음 페이지 처리 - 마지막 페이지인지 확인
           current_page_idx = metadata['page']
           if current_page_idx + 1 < len(pages):
               chunk_next_page = pages[current_page_idx + 1].page_content
           else:
               chunk_next_page = ""  # 마지막 페이지일 때 빈 문자열
           
           # 벌크 데이터에 추가
           bulk_data.append({
               "_index":  index_name,
               "_source": {
                   "title": book_name,
                   "total_pages": int(metadata.get('total_pages', 0)),
                   "page_number": int(metadata.get('page_label', 0)),
                   "chunk_text": text.page_content,
                   "chunk_id": i,
                   "chunk_page": pages[current_page_idx].page_content,
                   "chunk_next_page": chunk_next_page,
                   "chunk_embedding": embedding,
                   "chunk_embedding_1024": embedding_1024,
                   "author": author,
                   "source_file": pdf_file,
                   "book_id": book_id
               }
           })
   
   return bulk_data
```

```python
bulk_data=process_pdfs()
```

```python
if bulk_data:
    try:
        success, errors = helpers.bulk(client, bulk_data, raise_on_error=False, stats_only=False)
        print(f"성공: {success}")
        if errors:
            print(f"{len(errors)} documents failed to index.")
            for e in errors:
                print(e)
    except Exception as e:
        print(f"예외 발생: {e}")
 
else:
    print("인덱싱할 데이터가 없습니다.")
```
